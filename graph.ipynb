```python
import matplotlib.pyplot as plt
import json

filename = "Qwen_Qwen3-VL-4B-Instruct_qwen3vl_vllm_profile_summary.json"

    
with open(filename, 'r') as f:
    data = json.load(f)

experiments = [d["experiment"] for d in data]
mean_elapsed = [d["mean_elapsed_s"] for d in data]

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(experiments, mean_elapsed)
ax.bar_label(bars, fmt="%.2fs", padding=3)
ax.set_ylabel("Mean Elapsed (s) per image")
ax.set_title("Mean Elapsed Time per Image by Experiment (Qwen3-VL-4B-Instruct)")
plt.xticks(rotation=15, ha="right")
plt.tight_layout()
plt.show()



filename = "Qwen_Qwen3-VL-2B-Instruct_qwen3vl_vllm_profile_summary.json"

    
with open(filename, 'r') as f:
    data = json.load(f)

experiments = [d["experiment"] for d in data]
mean_elapsed = [d["mean_elapsed_s"] for d in data]

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(experiments, mean_elapsed)
ax.bar_label(bars, fmt="%.2fs", padding=3)
ax.set_ylabel("Mean Elapsed (s) per image")
ax.set_title("Mean Elapsed Time per Image by Experiment (Qwen3-VL-2B-Instruct)")
plt.xticks(rotation=15, ha="right")
plt.tight_layout()
plt.show()
```
